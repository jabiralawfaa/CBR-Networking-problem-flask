## Tambah Gejala dan Diagnosis Baru

```python
def tambah_gejala(nama_gejala_baru, data_gejala, data_diag):
    """Tambah satu gejala baru ke basis kasus.
    Semua kasus diagnosis lama otomatis diberi nilai 0 untuk gejala ini,
    supaya panjang vektor tetap sama dengan jumlah gejala."""
    nama_gejala_baru = nama_gejala_baru.strip().lower().replace(" ", "_")

    if nama_gejala_baru == "":
        print("Gagal: nama gejala tidak boleh kosong.")
        return False

    if nama_gejala_baru in data_gejala:
        print(f"Gagal: gejala '{nama_gejala_baru}' sudah ada (indeks {data_gejala.index(nama_gejala_baru)}).")
        return False

    data_gejala += [nama_gejala_baru]
    for baris in data_diag:
        baris += [0]

    print(f"Berhasil: gejala '{nama_gejala_baru}' ditambahkan (indeks {len(data_gejala) - 1}).")
    return True


def tambah_diagnosis(nama_baru, daftar_gejala, data_gejala, data_diag, nama_diag):
    """Tambah satu diagnosis (penyakit) baru beserta gejalanya.
    daftar_gejala berisi nama gejala. Gejala yang belum ada di data_gejala
    otomatis didaftarkan lebih dulu lewat tambah_gejala()."""
    nama_baru = nama_baru.strip()

    if nama_baru == "":
        print("Gagal: nama diagnosis tidak boleh kosong.")
        return False

    for nama in nama_diag:
        if nama.lower() == nama_baru.lower():
            print(f"Gagal: diagnosis '{nama_baru}' sudah ada.")
            return False

    if len(daftar_gejala) == 0:
        print("Gagal: diagnosis harus punya minimal satu gejala.")
        return False

    # daftarkan gejala yang belum ada
    gejala_bersih = []
    for g in daftar_gejala:
        g = g.strip().lower().replace(" ", "_")
        if g not in data_gejala:
            tambah_gejala(g, data_gejala, data_diag)
        if g not in gejala_bersih:
            gejala_bersih += [g]

    # buat deret biner untuk diagnosis baru
    deret_biner = [0] * len(data_gejala)
    for g in gejala_bersih:
        deret_biner[data_gejala.index(g)] = 1

    nama_diag += [nama_baru]
    data_diag += [deret_biner]

    print(f"Berhasil: diagnosis D{len(nama_diag)} | {nama_baru} ditambahkan dengan {len(gejala_bersih)} gejala.")
    return True


def buat_kasus_baru(daftar_gejala, data_gejala):
    """Ubah daftar nama gejala menjadi vektor biner kasus baru.
    Dipakai agar panjang vektor selalu mengikuti jumlah gejala terbaru."""
    kasus = [0] * len(data_gejala)
    for g in daftar_gejala:
        g = g.strip().lower().replace(" ", "_")
        if g in data_gejala:
            kasus[data_gejala.index(g)] = 1
        else:
            print(f"Peringatan: gejala '{g}' tidak dikenal, diabaikan.")
    return kasus


def tampilkan_basis_kasus(data_gejala, data_diag, nama_diag):
    print(f"Jumlah gejala: {len(data_gejala)} | Jumlah diagnosis: {len(nama_diag)}")
    print("=" * 60)
    for i in range(len(nama_diag)):
        gejala_terdeteksi = []
        for j in range(len(data_diag[i])):
            if data_diag[i][j] == 1:
                gejala_terdeteksi += [data_gejala[j]]
        print(f"D{i + 1} | {nama_diag[i]}")
        print("    Gejala: " + ", ".join(gejala_terdeteksi))
```
